```python
import os
from langchain.document_loaders import PyPDFDirectoryLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
CORPUS_PATH = "./zyro-dynamics-hr-corpus/" # Our Knoledge Base
loader =  PyPDFDirectoryLoader(CORPUS_PATH) # Load all PDF files in the directory
documents = loader.load()
print(f"Loaded {len(documents)} documents from {CORPUS_PATH}")
```

```python
spliter = RecursiveCharacterTextSplitter(
    chunk_size=800, 
    chunk_overlap=100
)
chunks = spliter.split_documents(documents)
print(f"Created {len(chunks)} chunks")
```
